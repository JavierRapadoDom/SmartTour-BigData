```python
from pathlib import Path
import pandas as pd
import os

carpeta = Path(r"C:\Users\alumnoCE\Documents\SmartTour-BigData\notebooks\Semana02\data")

archivos_atletico = sorted(carpeta.glob("atletico_*.csv"))


def cargar_jugadores(carpeta):
    jugadores = set()
    for archivo in os.listdir(carpeta):
        if archivo.endswith(".csv"):
            ruta = os.path.join(carpeta, archivo)
            df = pd.read_csv(ruta)

            # Añadimos jugadores al conjunto (set elimina duplicados automáticamente)
            jugadores.update(df["jugador"].dropna().astype(str))

    return jugadores
```
