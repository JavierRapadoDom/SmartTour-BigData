```python
#Creación de listas
camisetas = ["C1", "C2", "C3"]
pantalones = ["P1", "P2"]
zapatos = ["Z1", "Z2"]
```

```python
from itertools import product

outfits = list(product(camisetas, pantalones, zapatos)) #Nos da todas las combinaciones

#Imprimo el resultado
for outfit in outfits:
    print(outfit)
```

```python
print("Numero total de outfits: ", len(outfits))
```

```python
#añadir camiseta
camisetas.append("C4")

#recalculo combinaciones usando product de itertools
outfits_actualizado = list(product(camisetas, pantalones, zapatos))

print("Numero total de outfits: ", len(outfits_actualizado))
```

```python
#Representación en tabla:
import pandas as pd

df = pd.DataFrame(outfits_actualizado, columns=["Camiseta", "Pantalón", "Zapatos"])

print(df.head())
```
