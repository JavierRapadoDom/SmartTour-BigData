```python
grupo1 = {"Anuel", "Ernesto", "Lucas", "Marta   ", "Giuliano"}
grupo2 = {"Luciana", "Santiago", "Pedro", "Manuel", "Anuel"}

print(f"Grupo 1: {grupo1}")
print(f"Grupo 2: {grupo2}")

todos = grupo1 | grupo2
print(f"La unión de ambos grupos es: {todos}")

interseccion = grupo1 & grupo2
print(f"La intersección de ambos grupos es: {interseccion}")

diferencia = grupo1 - grupo2
print(f"La diferencia de ambos grupos es: {diferencia}")

diferencia_simetrica = grupo1 ^ grupo2
print(f"La diferencia simetrica de ambos grupos es: {diferencia_simetrica}")
```
