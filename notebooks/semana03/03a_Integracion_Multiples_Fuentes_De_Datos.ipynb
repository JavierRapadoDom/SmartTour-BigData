```python
import pandas as pd

### 1. Leer los ficheros (CSV de ventas, Excel de productos, JSON de opiniones)

## Ventas (CSV)
ventas = pd.read_csv("./Lab03a/ventas.csv")

## Productos (Excel)
productos = pd.read_excel("./Lab03a/productos.xlsx")

## Opiniones (JSON)
opiniones = pd.read_json("./Lab03a/opiniones.json")

### 2. Mostrar la informacion de los ficheros

# print(f"\n {ventas}")
# print(f"\n {productos}")
# print(f"\n {opiniones}")

### 3. Pequeño control de calidad de los datos(si hay valores nulos / si hay valores duplicados)

## Tipos de datos

# print(f"\n {ventas.info()}")
# print(f"\n {productos.info()}")
# print(f"\n {opiniones.info()}")

## Si hay valores nulos

# print("Valores nulos en ventas: ")
# print(ventas.isnull().sum())

# print("Valores nulos en productos: ")
# print(productos.isnull().sum())

# print("Valores nulos en opiniones: ")
# print(opiniones.isnull().sum())

## Si hay valores duplicados

# print("Valores duplicados en ventas: ")
# print(ventas.duplicated().sum())

# print("Valores duplicados en productos: ")
# print(productos.duplicated().sum())

# print("Valores duplicados en opiniones: ")
# print(opiniones.duplicated().sum())

### 4. Unificar los datos

## Ventas - Productos

ventas_productos = pd.merge(ventas, productos, on="producto_id", how="left") 
## En el on, la clave conjunta entre las dos tablas 
## En el how, cual de las dos tablas prevalece

# print(ventas_productos)

## Clientes (ventas_productos) - Opiniones

datos_completos = pd.merge(ventas_productos, opiniones, on="cliente_id", how="left")

# print(datos_completos)

### 5. Completamos informacion (total) / añadir columna total

datos_completos["total"] = (
    datos_completos["cantidad"] * datos_completos["precio"]
)

# print(datos_completos)
#

### 6. Extraccion de indicadores clave (KPIs) no lo hicimos / esta en el notion parte de ello

## Total gastado por cliente 

datos_completos["total"] = datos_completos["cantidad"] * datos_completos["precio"]

gasto_cliente = datos_completos.groupby("cliente_id")["total"].sum()
# print(gasto_cliente)

## Valoracion media por cliente

valoracion_media = datos_completos.groupby("cliente_id")["valoracion"].mean()
# print(valoracion_media)

## Producto más vendido
producto_popular = datos_completos["nombre"].value_counts()
print(producto_popular)
```
