```python
import pandas as pd

## Ventas (CSV)
clientes = pd.read_csv("./lab03b/clientes.csv")

## Productos (Excel)
ventas = pd.read_excel("./lab03b/ventas.xlsx")

## Opiniones (JSON)
contacto = pd.read_json("./lab03b/contacto.json")


# print(f"Clientes: \n{clientes}")
# print(f"Ventas: \n{ventas}")
# print(f"Contacto: \n{contacto}")

# Print antes de eliminar duplicados
print(f"Antes de eliminar duplicados: {len(clientes)}")

# Eliminar duplicados
clientes = clientes.drop_duplicates()


# Print después de eliminar duplicados
print(f"Después de eliminar duplicados: {len(clientes)}")

# Convertir el nombre para que la primera letra sea mayúscula
clientes["nombre"] = clientes["nombre"].str.strip().str.capitalize()

print(clientes["nombre"])


# Convertir todas las ciudades al mismo formato
clientes["ciudad"] = clientes["ciudad"].str.strip().str.title()
print(clientes["ciudad"])


# Convertir la fecha 
clientes["fecha_alta"] = pd.to_datetime(
    clientes["fecha_alta"], format = "%d/%m/%Y", errors="coerce"
).dt.strftime("%Y-%m-%d")

print(clientes["fecha_alta"])



clientes = clientes.rename(columns={"id_cliente": "cliente_id"})

# print(f"Clientes: \n{clientes}")
# print(f"Ventas: \n{ventas}")
# print(f"Contacto: \n{contacto}")


# INTEGRACIÓN 
integrado = clientes.merge(ventas, on="cliente_id", how="left")
integrado = integrado.merge(contacto, on="cliente_id", how="left")

print(integrado)


# --- Detectar incidencias ---  NO PRESENTAR
# ciudades = clientes["ciudad"].fillna("").str.strip()
# incidencias = {
#     "duplicados":      clientes.duplicated().sum(),
#     "ciudades_vacias": (ciudades == "").sum(),
#     "ventas_sin_cliente":      (~ventas["cliente_id"].isin(clientes["cliente_id"])).sum(),
#     "clientes_sin_correo":     (~clientes["cliente_id"].isin(contacto["cliente_id"])).sum(),
#     "contactos_sin_correspondencia": (~contacto["cliente_id"].isin(clientes["cliente_id"])).sum(),
# }

# print(incidencias)
```
